```python
def load_messages():
    messages = []
    file = 'database.json'

    empty = os.stat(file).st_size == 0

    if not empty:
        with open(file) as db_file:
            data = json.load(db_file)
            for item in data:
                messages.append(item)
    else:
        messages.append(
            {"role": "system", "content": f"""
                    You are a professional interviewer conducting a job interview. 
                    Use the candidate's resume and the job description to ask relevant questions.
                    Ask questions that assess the candidate's skills, experience, and fit for the role.
                    Ask one question at a time and wait for the response. 
                    Keep your tone simple and professional. Keep message medium length.
                    Based on the candidate's response, either:
                    1. Ask a follow-up question to dive deeper if the answer needs clarification or expansion
                    2. Move on to a new question about a different aspect of their experience or skills
                    
                    Resume: {resume_text}
                    
                    Job Description: {job_description}"""
            }
        )
    return messages
```

```python
def save_messages(user_message, gpt_response):
    file = 'database.json'
    messages = load_messages()
    messages.append({"role": "user", "content": user_message})
    messages.append({"role": "assistant", "content": gpt_response})
    with open(file, 'w') as f:
        json.dump(messages, f)
```

```python
def get_chat_response(user_message):
    messages = load_messages()
    messages.append({"role": "user", "content": user_message})
    

    # Send to ChatGpt/OpenAi
    model_response = model.invoke(messages)

    parsed_model_response = model_response.content

    # Save messages
    save_messages(user_message, parsed_model_response)

    return parsed_model_response
```

```python
if __name__ == "__main__":
    print("AI: Give a Brief Introduction about yourself.")
    while True:
        messages = load_messages()
        if len(messages) > 10:
            print("Thank you for your time. Wait for the analysis...")
            break
        
        answer = input("You:")
        print(f"You: {answer}")
        chat_response = get_chat_response(answer)
        print(f"AI: {chat_response}")
        speaker.yap(chat_response)
```
